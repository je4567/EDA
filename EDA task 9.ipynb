```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("kc_house_data.csv")
df
```

```python
df.columns
```

```python
print("missing values")
print(df.isna().sum())
```

```python
print("Zero values in bedrooms:",(df["bedrooms"]==0).sum())
print("Zero values in bathrooms:",(df["bathrooms"]==0).sum())
```

```python
df["bedrooms"]=df["bedrooms"].replace(0,np.nan)
df["bathrooms"]=df["bathrooms"].replace(0,np.nan)
```

```python
df.isna().sum()
```

```python
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].mean())
df['bathrooms'] = df['bathrooms'].fillna(df['bathrooms'].median())
```

```python
df.isna().sum()
```

```python
df["date"]=pd.to_datetime(df["date"])
```

```python
df['sale_year']=df["date"].dt.year
```

```python
df['house_age']=df['sale_year']-df["yr_built"]
```

```python
df["years_since_renovation"]=np.where(
    df["yr_renovated"]>0,
    df['sale_year']==df["yr_renovated"],
    np.nan
)
```

```python
df[["date","sale_year","yr_built","yr_renovated","years_since_renovation"]].head()
```

```python
df["price_per_sqft_living"]=df["price"]/df["sqft_living"]
df["price_per_sqft_lot"]=df["price"]/df["sqft_lot"]
```

```python
df[["price","sqft_living","sqft_lot","price_per_sqft_living","price_per_sqft_lot"]].head()
```

```python
plt.figure(figsize=(9,5))
sns.boxplot(x=df["price"])
plt.title("box plot of house price")
plt.xlabel("price")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.boxplot(x=df["price_per_sqft_living"])
plt.title("box plot of price_per_sqft_living")
plt.xlabel("price_per_sqft_living")
plt.show()
```

```python
Q1=df["price"].quantile(0.25)
Q3=df["price"].quantile(0.75)

IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("lower_bound:",lower_bound)
print("upper_bound:",upper_bound)
```

```python
print_outliers=df[
    (df["price"]<lower_bound)|
     (df["price"]>upper_bound)]
print("number of price outliers:",len(print_outliers))
print_outliers[["id","price","bedrooms","bathrooms","sqft_living"]].head(20)
```

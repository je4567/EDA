```python
df["Date of Admission"]=pd.to_datetime(df["Date of Admission"])
```

```python
df["Discharge Date"]=pd.to_datetime(df["Discharge Date"])
```

```python
df.info()
```

```python
#scatterplot
df_pivot=pd.pivot_table(
    df,
    values="Billing Amount",
    index="Medical Condition",
    columns="Insurance Provider",
    aggfunc="sum",
)
df_pivot
```

```python
df_pivot.plot(kind="bar",stacked=True,figsize=(12,6))
plt.title("Billing Amount by Medical Condition and Insurance Provider")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
sns.violinplot(x="Insurance Provider",y="Billing Amount",data=df)
plt.title("Billing Amount by Insurance Provider")
plt.xlabel("Insurance Provider")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
sns.lineplot(x="Date of Admission",y="Discharge Date",data=df)
plt.title("Date of Admission and Discharge Date")
plt.xlabel("Date of Admission")
plt.ylabel("Discharge Date")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
df["Hospital stay days "]=(df["Discharge Date"]-df["Date of Admission"]).dt.days
```

```python
numeric_column=["Age","Hospital stay days","Billing Amount"]
correlation=df[numeric_column].corr()
correlation
```

```python
sns.heatmap(correlation,
            annot=True,
            cmap="Reds",
            cbar=False,
            square=True,
            linewidth=1)
plt.title("correlation of age,hospital stay days and billing  amount")
plt.show()
            
            
```
